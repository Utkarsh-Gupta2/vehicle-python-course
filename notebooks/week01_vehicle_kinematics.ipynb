```python
# Create a time vector from 0 to 10 seconds with 100 data points
start_time = 0
end_time = 10
num_points = 100
t = np.linspace(start_time, end_time, num_points)

# Print the array (first few values and last few values)
print(t[:5])
print("...")
print(t[-5:])
```

```python
# Constant Acceleration Model
initial_velocity = 0  # meters per second
acceleration = 2  # meters per second squared   

# Calculate velocity over time
velocity = initial_velocity + acceleration * t
print(velocity[:5]) 
print("...")
print(velocity[-5:])        

#Calculate position over time
# Integrate velocity to get displacement
# 'x=t' specifies the time values for integration
# 'initial=0' sets the starting displacement at time 0 to zero
displacement = cumulative_trapezoid(velocity, x=t, initial=0)
print(displacement[:5])
print("...")
print(displacement[-5:])
```

```python
# Plotting the results
plt.figure(figsize=(8, 4))
plt.plot(t, velocity, 'o-', label='Velocity (m/s)')
plt.plot(t, displacement, 's-', label='Displacement (m)')
plt.xlabel('Time (s)')
plt.ylabel('Value')
plt.title('Velocity and Displacement over Time')
plt.legend()
plt.grid(True)
plt.show()
```
